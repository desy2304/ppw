# Klasifikasi Berita Menggunakan Naive Bayes

Setelah dilakukan proses TF-IDF untuk mengubah data teks menjadi representasi numerik dan reduksi dimensi data dengan PCA, tahap selanjutnya adalah melakukan proses klasifikasi menggunakan algoritma Naive Bayes.

Algoritma Naive Bayes digunakan untuk menentukan kategori dari setiap data berdasarkan pola yang dipelajari dari data training. Pada tugas ini, model melakukan klasifikasi ke dalam dua kelas, yaitu Sport dan Finance.

Sebelum proses klasifikasi, dataset dibagi menjadi data training dan data testing dengan perbandingan 80:20. Dari total 200 data, diperoleh 160 data training (80%) dan 40 data testing (20%). Data training digunakan untuk membangun model Naive Bayes, sedangkan data testing digunakan untuk mengukur kemampuan model dalam mengklasifikasikan data yang digunakan selama proses pelatihan

# Proses Klasifikasi Naive Bayes Menggunakan Orange

## 1. Import Dataset ke Orange

Langkah pertama adalah memasukkan dataset ke dalam Orange menggunakan widget CSV File Import.

![Image.png](images/Screenshot-2026-09-23-224237.png)

## 2. Memilih Kolom Data

Setelah dataset berhasil dimuat, tambahkan widget Select Column. Lalu hubungkan CSV file import ke select columns.

![image-2.png](images/Screenshot-2026-09-23-224334.png)

Di dalam widget select columns pilih semua features PC1-PC144, meta id serta targetnya adalah label/kategori.

## 3. Membagi Dataset dengan Data Sampler

Dataset kemudian dibagi menjadi data training dan data testing menggunakan widget Data Sampler. Hubungkan select columns dengan data sampler.

![image.png](images/Screenshot-2026-09-23-224443.png)

Pada data sampler, gunakan pengaturan sampling type fixed proportion of data, dengan proportion 80%.

## 4. Membuat Model Naive Bayes

Selanjutnya tambahkan widget Naive bayes. Kemudian hubungkan data sampler dengan model naive bayes dengan koneksi data sample

![image.png](images/Screenshot-2026-09-23-224815.png)

## 5. Evaluasi Model

Tambahkan widget Test & Score. Hubungkan data sampler dan test & score dengan dua koneksi yaitu, koneksi data sample dengan data dan koneksi remaining data dengan test data.

Hubungkan juga model naive bayes dengan test & score

![image.png](images/Screenshot-2026-09-23-225324.png)


![image-2.png](images/Screenshot-2026-09-23-225406.png)

Dalam konfigurasi test & score, pilih test on test data.

Dengan konfigurasi tersebut, model naive bayes dilatih menggunakan 160 data training, kemudian diuji menggunakan 40 data testing.

Berdasarkan hasil test & score, algoritma naive bayes memperoleh accuracy sebesar 85% dengan F1 0.850 ; Prec 0.855 ; recall 0.850
