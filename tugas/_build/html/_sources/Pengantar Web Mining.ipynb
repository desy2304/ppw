# Pengantar Web Mining

Web mining merupakan proses untuk menemukan dan mengekstrak informasi atau pengetahuan yang bermanfaat dari data yang tersedia di web. Web mining memanfaatkan berbagai teknik dari data mining, tetapi memiliki karakteristik yang berbeda karena data pada web sangat besar, kompleks, dinamis, serta memiliki format yang beragam seperti HTML, XML, dan teks.

Secara umum, web mining dapat dikelompokkan menjadi tiga jenis, yaitu web content mining dan web structure mining.

1. Web Content Mining
Web Content Mining berfokus pada pengambilan informasi dari isi halaman web, seperti teks, gambar, audio, video dan tabel. Contohnya adalah klasifikasi dokumen, analisis sentimen, ekstraksi informasi dan peringkasan teks.

2. Web Usage Mining
Web Usage Mining bertujuan menemukan pola perilaku pengguna berdasarkan aktivitas di website, seperti server log, cookie, dan clickstream. Hasilnya dapat digunakan untuk membuat rekomendasi atau memahami kebiasaan pengguna.

3. Web Structure Mining
Web Structure Mining mempelajari struktur dan hubungan antarhalaman web melalui hyperlink. Metode ini dapat digunakan untuk menemukan halaman penting, menganalisis jaringan dan mendeteksi komunitas. Salah satu contohnya adalah PageRank.

# Proses Web Mining

Secara umum, proses Web Mining memiliki tahapan yang mirip dengan data mining. Perbedaannya terdapat pada proses pengumpulan dan pemrosesan data web.

Tahapan tersebut meliputi:

1. Gathering dan Exploration
Mengumpulkan data dari web melalui crawling atau Web API, kemudian memahami dan mengeksplorasi data yang diperoleh.

2. Preprocessing dan Transformation
Mempersiapkan data agar sesuai dengan metode yang akan digunakan. Tahap ini dapat mencakup seleksi fitur, reduksi dimensi, transformasi teks menjadi vektor atau embedding, serta penggabungan data dari berbagai sumber.

3. Data Mining
Menerapkan metode data mining pada data yang telah diproses untuk menghasilkan model atau pola. Hasilnya kemudian dievaluasi dan dapat dilakukan eksperimen atau perbaikan secara berulang.
